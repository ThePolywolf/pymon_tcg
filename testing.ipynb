```python
pymon_v = pl.read_parquet("./data/encoded/pymon.vectors.parquet")
pymon.head(5)
```

```python
errors = 0
for row in pymon_v.iter_rows():
    id = row[0]
    sel = pymon.filter(pl.col('id') == id)
    count = sel['id'].count()

    if count == 0:
        errors += 1
        print(f"Id {id} has no matches (out of bounds?)")
    
    if count > 1:
        errors += 1
        print(f"Id (id) has {count} collisions (1 expected)")

if errors == 0:
    print("All tests passed, no collisions or failures")
```
